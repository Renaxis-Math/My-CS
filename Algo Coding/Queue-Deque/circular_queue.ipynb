## Invariant

1. The whole point of implementation is to not use `push_back()` (C++) or `.append()` (Python)

2. The status of the queue is represented by 2 indices: *back*, *front* and 1 variable: *size*


    1. *front_i* and *back_i* represents elements existing in the queue.
    2. this means new element is inserted in `_wrap_around(back_i + 1)` index.
    3. similarly when `pop()`, `front_i` is moved in `_wrap_around(front_i + 1)` manner.
    4. we need `size` variable as `front_i` and `back_i` can be anywhere.

## Circular Queue

![](../../.images/circular_queue.png)

```cpp
class MyCircularQueue {

private:
    int back_i, front_i;
    vector<int> queue;      // will be init with 'capacity'
    int size;
public:
    MyCircularQueue(int capacity) : 
        back_i{-1}, front_i{0}, queue(capacity), size{0}
    {}
    
    bool push_back(int value) {
        if (is_full()) return false;
        back_i = _wrap_around(back_i + 1);
        queue[back_i] = value;
        size++;
        return true;
    }
    
    bool pop() {
        if (is_empty()) return false;
        front_i = _wrap_around(front_i + 1);
        size--;
        return true;
    }
    
    int front() { return is_empty() ? -1 : queue[front_i]; }
    
    int back() { return is_empty() ? -1 : queue[back_i]; }
    
    bool is_empty() { return size == 0; }
    
    bool is_full() { return size == queue.size(); }

    int _wrap_around(int new_i) {
        int capacity = queue.size();
        return ((new_i % capacity) + capacity) % capacity;
    } 
};
```
