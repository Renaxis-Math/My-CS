## Tail Recursion

1. The thing that makes entire function "tail recursive" is that the recursion is made **only in the last statement of a function**. If it appears elsewhere, the entire function is "regular" recursion.
```cpp
    ... f():
        // do something non-recursive
        return f() // or just f(), but not f() + f()
```


2. Not all functions are tail recursive because sometimes you need to perform the recursion in the middle of the function.


3. A "tail recursive" function applies Tail Call Optimization (TCO), a compiler thing that **reuses the current function's stack frame instead of adding new frames** for subsequent recursive calls, *reduce memory complexity from O(n) to O(1)*


4. C++'s TCO is not auto. To invoke TCO when compiling in C++, must use
`g++ -O2`

5. In tail recursion, assuming we want to return something, it's mostly true that in base case we will return something related to the arguments, while in normal case, we return the recursive function call. Look at these examples below.

### Example 1: Fibonacci

````python
def fib_tail(n, prev, cur):
    if n == 0: return prev
    if n == 1: return cur
    return fib_tail(n - 1, cur, prev + cur)

def fib(n):
    return fib_tail(n, 0, 1)

assert fib(4) == 3
````

### Example 2: gcd(x, y)

````python
def gcd(x, y):
    if x == 0: return y
    return gcd(y % x, x)
````

## Tail Recursion = While loop

It's easy to switch by:
1. Base case return values = starting values
2. `while` loop until base case condition:
    - argument changes = update *existing* variables
3. Return the variable based on the base case that `while` loop stops

### Example 1: Fibonacci

````python
# Base case stops at n == 0
def fib0(n):
    prev, cur = 0, 1
    while (n):
        prev, cur = cur, prev + cur
        n -= 1
    return prev

# Base case stops at n == 1
def fib1(n):
    if n == 0: return 0

    prev, cur = 0, 1
    while (n > 1):
        prev, cur = cur, prev + cur
        n -= 1
    
    return cur

for n in range(0, 31):
    assert fib0(n) == fib1(n)
````

### Example 2: gcd

````python
def gcd(x, y):
    if x == 0: return y
    return gcd(y % x, x)

def gcd_iterative(x, y):
    while x:
        x, y = y % x, x
    return y

for x in range(1, 20):
    for y in range(1, 20):
        assert gcd(x, y) == gcd_iterative(x, y)
````
