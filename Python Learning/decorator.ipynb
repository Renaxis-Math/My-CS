1. Decorator of a `func` = input is `func`, output is `func`.
2. Decorator expressions are evaluated when the function is defined.
3. The result must be a callable, which is invoked with the function object as the only argument. 

```python
def log_calls(fn):
    def wrapped(*args, **kwargs):
        print("calling", fn.__name__)
        return fn(*args, **kwargs)
    return wrapped
```

How to use it

```python
@log_calls
def add(a, b):
    return a + b
```

Now `add(2, 3)` actually calls `wrapped`, which prints something, then calls the original `add`.
