A generator function is a function that can pause and resume, instead of computing everything at once.

In Python, it is usually a function that uses `yield`.

```python
def f():
    print("A")
    x = 10
    yield x
    print("B")
    y = x + 5
    yield y
    print("C")
```

1. When the above function being called, nothing will run.

```python
g = f() # g is an iterator
```

2. The function only runs when calling `next(g)`. The expressions will run until:
- Meeting a `yield ...`, returning `...`, function can resume in later `next(g)` call.
- Meeting a `return ...` or end of function, function STOPS, returning a `StopIterator`

```python
g = f()        # does not run body yet

next(g)        # prints A, yields 10
next(g)        # prints B, yields 15
next(g)        # prints C, then StopIteration
```
