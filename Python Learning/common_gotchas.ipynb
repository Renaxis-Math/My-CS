## Default arg

Principles:
- anything inside function body is binded only when function called
- default argument is evaluated when function is created i.e `def`

```python
def appendNew(appendTo =[]):      
    appendTo.append(1)   
    return appendTo

print(appendNew())
print(appendNew())
```

## Late binding

Same principles:
- anything inside function body is binded only when function called
- default argument is evaluated when function is created i.e `def`

```python
def create_multipliers():
    return [lambda x: i * x for i in range(5)]

for multiplier in create_multipliers():
    print(multiplier(2))
```

`i` inside the function body is looked up when the function is called.

The `for i in range(5)` runs first. After the loop finishes, `i` is 4. 

So every function you created looks up `i`, finds 4, and uses that

That code is equivalent to:

```python
def create_multipliers():
    multipliers = []
    for i in range(5):
        def multiplier(x): return i * x
        multipliers.append(multiplier)
    return multipliers

for multiplier in create_multipliers():
    print(multiplier(2))
```

The fix is to put `i` in function **argument**. 

A default argument like `i=i` is evaluated when the function is created.

```python
def create_multipliers():
    return [lambda x, i=i: i * x for i in range(5)]
```

## Parenthesis

```python
print(5>2 == True)
```

This results in False because the above expression effectively means that `5 > 2 & 2 == True` (same thing with `1 <= x <= 2` means `1 <= x & x <= 2`) 

This implies, `True & False`. Hence, the final result is False.

```python
print(5 is (not None)) # equals to 5 is True, which is always False
print(5 is not None)
```

`+=` means re-assign means "change the scope"

```python
list1 = [1, 2, 3]

def baz1():
    # the code works fine
    list1.append(4) 
    return list1

def baz2():
    # Doesn't work fine
    list1 = list1 + [5]      
    return list1
```

```python
print(baz1()) # list1 is still 'global'
```

```python
print(baz2()) # list1 becomes 'local', and it's not init yet
```

There are 2 ways to fix:

1. use keyword `global`

```python
list1 = [1, 2, 3]

def baz2_fix1():
    global list1
    list1 += [5]
    return list1

baz2_fix1()
```

2. Apply "default arg" principle

```python
list1 = [1, 2, 3]

def baz2_fix2(lst):
    lst += [5]
    return lst

baz2_fix2(list1)
```

## Mutating a list while iterating over it

While working with a list or other mutable items, if we mutate it while iterating over it, it's certain to cause errors.

This will cause an error

```python
nums = [_ for _ in range(10)]
for i in range(len(nums)):
    if nums[i] % 2 != 0:
        del nums[i]
print(nums)
```

Same principle: this will cause an infinite loop

```python
nums = [_ for _ in range(10)]
for num in nums: # for i in range(len(nums)) works
    if num % 2 != 0:
        nums.append(1)
print(nums)
```

If using `for i, num in enumerate(nums)`, `.append()` and `del` will cause similar issues, but modifying values won't

```python
nums = [1, 2, 3]
for i, num in enumerate(nums):
    nums[i] = num + 100
```
