## Template is not universal

```cpp
template <typename T>
class Box {
public:
    Box(T val) : value{val} {}
    T getValue(); // Declaration only
private:
    T value;
};

// Definition outside the block
template <typename T>
T Box<T>::getValue() {
    return value;
}
```

- `template <typename T>`: You must tell the compiler again as the function is out

- `Box<T>::`: Notice you can't just write `Box::`. You must include the `<T>` because Box isn't a class; `Box<T>` is the template class.

## Template "Header file" Trap

1. if i use template in a header file, that function must have full implementation.

    ```cpp
    // Box.h
    template <typename T>
    class Box {
    public:
        // Full implementation right here!
        T getValue() { 
            return value; 
        }
    private:
        T value;
    };
    ```

2. If template inside a .cpp file, use it for every function / class using T


## Comparison Template can be confusing

1. We can always use explicit instantiation in ambiguous cases like this. These 2 belows produce the same result.

    ```cpp
    template<typename T>
    T min(const T& a, const T& b) {
        return a < b ? a : b
    }

    std::cout << min<double>(106, 3.14);
    ```

    ```cpp
    template<typename T, typename U>
    auto min(const T& a, const U& b) {
        return a < b ? a : b
    }

    std::cout << min(106, 3.14);
    ```

2. T must have a comparison `operator<`

What if template comparison function but T doesn't have `operator<` ?

## Concept

Concept greatly reduces compile error message.

```cpp
template <typename T>
concept Comparable = requires(T a, T b) {
    { a < b } -> std::convertible_to<bool>;
};

// Super slick shorthand for the above
template <Comparable T>
T min(const T& a, const T& b);
```

## Varying Args Template

It's like recursion:
1. Base case
2. Normal recursive calls

```cpp

/* Base case */
template <Comparable T>
T min(const T& v) { return v; }

/* Normal case */
template <Comparable T, Comparable... Args>
T min(const T& v, const Args&... args) {
    auto m = min(args...);
    return v < m ? v : m;
}
```
