## Function Overload in Argument Type is Allowed

```cpp
double f(int x) return (double)x + 3;
double f(double x) return x * 3;
```

## Struct

1. Main idea is to return 1 group of things together
2. Elements are **public** by default. But can be *private* or *protected*

### . or ->

```cpp
struct Person p1;
p1.age = 25; // Direct access
```

```cpp
struct Person *ptr = &p1;
ptr->age = 30; // Shorthand for (*ptr).age
```
