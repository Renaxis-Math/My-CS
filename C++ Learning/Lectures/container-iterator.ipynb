## Vector

### [] does not do bound checking

```cpp
std::vector<int> vec{5, 6}; // {5, 6}
vec[1] = 3; // {5, 3}
vec[2] = 4; // undefined behavior
vec.at(2) = 4; // Runtime error
```

## Unordered-Map

Unordered-Map is just a container of associated pairs

```cpp
std::map<std::string, int> map {
    { "Chris", 2 },
    { "CS106L", 42 },
    { "Keith", 14 },
    { "Nick", 51 },
    { "Sean", 35 },
};

int sean = map["Sean"]; // 35
map["Chris"] = 31;
```

### Hash Function

- returns a `size_t` (64 bits)
- Small changes in the input should produce large changes in the output
- If two keys hash to the same bucket, we get a hash collision
- Load factor: average number items per bucket. If it's too huge, we must re-write hash function

## Map vs Set

- Ordered-Map is a binary search (red-black) tree where Node = <Key, Val>
- Ordered-Set is a binary search (red-black) tree where Node is just Key, no Val
