## Uniform Initialization
Uniform Initialization triggers a compile error if data type not matched

```cpp
std::vector<int> v{10, 20}; // 2 elements: 10, 20
std::vector<int> v(10, 20); // 10 elements, init with all 20
std::vector<pair<int, int>> v{{10, 20}}; // 1 element, the pair {10, 20}
```

## Using `auto [a, b] = ...`

1. Only applicable for receiving variables with **fixed-at-compile-time** number of elements. E.g:
    - pair
    - tuple
    - array
    - struct

2. This creates a **copy**, if you want [references](./function.ipynb), use `auto&`
