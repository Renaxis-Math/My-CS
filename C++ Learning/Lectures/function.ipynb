## Pass by reference

1. Pass-by-Ref. function argument name $\neq$ from variable name, due to scope

    ```cpp
    void f(int& a) { a += 1; } // function scope
    int a = 5;                   // outer scope
    ```

2. Beware of `auto&` vs `auto`:

    - Not modify the pair in-place:
        ```cpp
        void f(vector<pair<int, int>>& nums) {
            for (auto [num1, num2] : nums) num1++;
        }
        ```
    - Modify the pair in-place:
        ```cpp
        void f(vector<pair<int, int>>& nums) {
            for (auto& [num1, num2] : nums) num1++;
        }
        ```    

## l-value and r-value

![](../../_images/lvalue_rvalue.png)

1.  rvalue type = no identity / memory address
    lvalue type = has identity (variable name) / memory address

2. Tricks:

    - T& binds to lvalue
    - const T& binds to lvalue or rvalue
    - T&& binds to rvalue

    ```cpp
    int x = 5;

    int& a = x;        // OK
    const int& b = 5;  // OK
    int&& c = 5;       // OK

    // NOT OK: non-const lvalue binds to 5, a r-value
    int& d = 5;        
    ```
