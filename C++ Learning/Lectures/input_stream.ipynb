## Read file

### Read line-by-line
```cpp
#include<fstream>
ifstream input_stream(my_file_name);

string s;
input_stream >> s;  // reads one word, stop at whitespace

/* Continue Reading from what's currently pointed */
vector<string> lines;
string line;
while (getline(input_stream, line)) // reads one whole line
    lines.emplace_back(line);

```

## Cout

Use `std::ios::sync_with_stdio(false)` for performance boost. BUT:

**This only works if your output stream is non-interactive!**

it only stopped flushing `‘\n’`s when the output
stream was non-interactive (i.e. file, Unix pipe).


However, if the output stream was interactive (i.e.
terminal), the output stream still interpreted it as a
line buffer, resulting in an immediate flush when `‘\n’`
was pushed to the stream.

### `cout << ... << flush;` vs `cout << ..;`

There is a difference.

`cout << "A";`

- writes to buffer
- may appear immediately, or may wait

`cout << "A" << flush;`

- writes to buffer
- then forces it to appear now

So `flush` matters when timing matters. In many simple terminal cases, they look the same, which is why it feels like no difference.
